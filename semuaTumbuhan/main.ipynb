```python
import os
import glob
import pandas as pd
```

```python
df_salada = pd.read_excel('../data_final_salada.xlsx')
df_pakcoy = pd.read_excel('../data_final_pakcoy.xlsx')
```

```python
print(df_salada.info())
```

```python
print(df_pakcoy.info())
```

```python
df_salada['plant_type'] = 'Lettuce'
df_pakcoy['plant_type'] = 'Pak Choi'
```

```python
df_combined = pd.concat([df_salada, df_pakcoy], ignore_index=True, sort=False)
```

```python
if 'leaf_width' in df_combined.columns:
    df_combined['leaf_width'].fillna(df_combined['leaf_width'].mean(), inplace=True)
```

```python
df_combined.to_excel("data_final_semua.xlsx", index=False)
```
